️**1) How many customers actually left?**

```python
df['exited'].value_counts()      

# 0 = Customer stayed
# 1 = Customer exited
```

```python
df['exited'].value_counts(normalize=True) * 100
```

**Result :- 2,037 out of 10,000 customers have exited, resulting in a 20.37% churn rate.
This indicates that approximately 1 in 5 customers has left the bank.**

**2) Which geography has the highest customer churn?**

```python
df.groupby('geography')['exited'].mean()*100
```

**Result:- Germany has the highest churn rate at 32.44%, nearly twice the rate of France and Spain. This makes German customers a key segment for retention efforts.**

**3) Are inactive customers more likely to leave?**

```python
df.groupby('is_active_member')['exited'].mean()*100
```

**Business Insight:-
Inactive customers have a much higher churn rate (26.87%) than active customers (14.27%), suggesting that low customer engagement is strongly associated with churn.**

**4) Does the number of products affect churn?**

```python
df.groupby('num_of_products')['exited'].mean()*100
```

```python
df['num_of_products'].value_counts().sort_index()
```

**Business Insight:-
Customers with 1 product have a relatively high churn rate (27.71%), while 2-product customers have much lower churn (7.60%). The 3- and 4-product groups show extremely high churn but contain only 326 customers, so they require further investigation before drawing conclusions.**

**5) Which age group has the highest churn?**

```python
df['age_group'] = pd.cut(df['age'],
                  bins=[18, 29, 39, 49, 59, 100],
                  labels=['18-29', '30-39', '40-49', '50-59', '60+'])

df['age_group']
```

```python
df['age_group'].value_counts().sort_index()
```

```python
df.groupby('age_group',observed=True)['exited'].mean()*100
```

**Business Insight
Customers aged 50–59 have the highest churn rate at 56.04%, making this the most critical age segment for retention efforts.**

**6) Does customer satisfaction affect churn?**

```python
df['satisfaction_score'].value_counts().sort_index()
```

```python
df.groupby('satisfaction_score')['exited'].mean()*100
```

**Business Insight:-
Churn rates remain around 20% across all satisfaction levels, suggesting that satisfaction score alone is not a strong predictor of churn in this dataset.**

**7) Does having a credit card affect churn?**

```python
# 0 = No credit card
# 1 = Has credit card
```

```python
df.groupby('has_credit_card')['exited'].mean()*100
```

**Business Insight:-
Churn is nearly identical for customers with and without a credit card, suggesting credit-card ownership has little impact on churn in this dataset.**

**8) How does customer tenure relate to churn?**

```python
df['tenure'].value_counts().sort_index()
```

df.groupby('tenure')['exited'].mean()*100

**Business Insight:-
Churn varies moderately across tenure groups, but no clear relationship is visible. Customers with 7 years of tenure have the lowest churn at 17.22%.**

**9) Does having more bank products affect customer churn?**

```python
df.groupby('num_of_products')['exited'].mean()*100
```

```python
df['num_of_products'].value_counts().sort_index()
```

**Business Insight:-
Customers with 2 products have the lowest churn rate at 7.60%, while churn rises sharply to 82.71% for customers with 3 products and reaches 100% for customers with 4 products. However, the 3- and 4-product groups are relatively small, with 266 and 60 customers respectively, so these extreme churn rates should be interpreted cautiously.**

**10) How does customer activity status relate to churn?**

```python
df.groupby('is_active_member')['exited'].mean()*100
```

```python
df['is_active_member'].value_counts().sort_index()
```

**Business Insight:-
Inactive customers have a significantly higher churn rate (26.87%) compared with active customers (14.27%), a difference of 12.60 percentage points. This suggests that customer engagement is strongly associated with retention, making inactive customers an important target for retention campaigns.**

**11)What percentage of customers have churned versus stayed with the bank?**

```python
sns.countplot(data=df,x='exited',color='skyblue',edgecolor='black',width=0.5)

plt.title('Customer Churn Distribution')

plt.xlabel('Exited')

plt.ylabel('Number of Customers')

plt.show()
```

**EDA Insight:-
The dataset has an overall churn rate of 20.38%, meaning approximately 1 in 5 customers have exited the bank.**

**12)How are customers distributed across different age groups?**

```python
plt.figure(figsize=(8,5))

sns.histplot(df['age'],bins=20,color='orange')

plt.title("Distribution of Customer Age")

plt.xlabel("Age")

plt.ylabel("Number of Customers")

plt.show()
```

**EDA Insight:-
This graph shows how customers are distributed across different age ranges.**

**Are there any unusual or extreme values in important numerical customer variables?**

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['age'])

plt.title('Age Distribution and Outliers')
plt.xlabel('Age')

plt.show()
```

**EDA Insight:-
Age contains several statistical outliers above approximately 62 years, extending up to 92 years. These values are plausible customer ages, so they should be retained rather than treated as data errors.**

**Which geography has the highest churn?**

```python
churn_geo = df.groupby('geography')['exited'].mean().mul(100)

plt.figure(figsize=(6,4))

plt.bar(churn_geo.index,churn_geo.values,color='skyblue',edgecolor='red',width=0.7)

plt.title("Customer Churn Rate By Geography")

plt.xlabel("Geography")

plt.ylabel("Churn Rate")

plt.show()


# I wanted to show you another Pandas method.
# Personally, I still prefer *100 instead of mul(100)
```

**Business Insight:-
Germany records the highest churn rate (32.44%), nearly double that of France and Spain, making it the highest-priority geography for customer retention initiatives.**

**Which age group has the highest customer churn?**

```python
churn_age = df.groupby('age_group',observed=True)['exited'].mean()*100

plt.figure(figsize=(6,4))

plt.bar(churn_age.index,churn_age.values,color='pink',edgecolor='blue')

plt.title("Customer Churn By Age Group")

plt.title("Customer Churn Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Churn Rate (%)")

plt.show()
```

**Business Insight:-
Customers aged 50–59 have the highest churn rate (56.04%), while customers aged 18–29 have the lowest churn (7.54%). This suggests that middle-aged customers should be prioritized in retention strategies.**
